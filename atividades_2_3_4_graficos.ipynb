```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Atividade 2 — Horas que mulheres e homens dedicam aos afazeres domésticos

```python
indicador_1 = pd.read_excel("dados/atividade_2_horas_domesticos.xlsx")

regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

reg = indicador_1[indicador_1["uf_regiao"].isin(regioes)].copy()

reg_long = reg.melt(
    id_vars="uf_regiao",
    value_vars=["homens", "mulheres"],
    var_name="sexo",
    value_name="horas",
)

reg_long
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=reg_long, x="uf_regiao", y="horas", hue="sexo", ax=ax)
ax.set_title("Afazeres domésticos por região e sexo")
ax.set_xlabel("Região")
ax.set_ylabel("Horas por semana")
plt.tight_layout()
plt.show()
```

## Atividade 3 — Formação acadêmica

```python
tabela3 = pd.read_excel("dados/tabela_10063_dados.xlsx")

formacao_total = tabela3.rename(columns={
    "Região": "uf_regiao",
    "Curso superior - Homens": "total_homens",
    "Curso superior - Mulheres": "total_mulheres"
})[["uf_regiao", "total_homens", "total_mulheres"]]

formacao_stem = tabela3.rename(columns={
    "Região": "uf_regiao",
    "CTEM - Homens": "homens",
    "CTEM - Mulheres": "mulheres"
})[["uf_regiao", "homens", "mulheres"]]

formacao_saude = tabela3.rename(columns={
    "Região": "uf_regiao",
    "Educação/Serviços pessoais/Saúde e Bem-estar - Homens": "homens",
    "Educação/Serviços pessoais/Saúde e Bem-estar - Mulheres": "mulheres"
})[["uf_regiao", "homens", "mulheres"]]
```

```python
brasil = pd.DataFrame(
    {
        "area": ["Graduação (total)", "STEM", "Saúde / educação"],
        "homens": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_homens"].iloc[0],
            formacao_stem.loc[formacao_stem["uf_regiao"] == "Brasil", "homens"].iloc[0],
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "homens"].iloc[0],
        ],
        "mulheres": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_mulheres"].iloc[0],
            formacao_stem.loc[formacao_stem["uf_regiao"] == "Brasil", "mulheres"].iloc[0],
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "mulheres"].iloc[0],
        ],
    }
)

brasil_long = brasil.melt(
    id_vars="area",
    var_name="sexo",
    value_name="pessoas"
)

brasil_long
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```

## Atividade 4 — População 2022 por UF

```python
pop_estado = pd.read_excel("dados/atividade_4_populacao_2022_uf.xlsx")

pop_estado.head()
```

```python
ordem = pop_estado.sort_values("2022", ascending=False)

fig, ax = plt.subplots(figsize=(9, 8))
sns.barplot(data=ordem, y="UF", x="2022", ax=ax, color="steelblue")
ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
